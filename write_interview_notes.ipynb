```python
import pathlib

content = r'''# Interview notes — self-correction on this project

Local file. Not committed, not part of the repo. This is raw material for
"tell me about a time you found an error in your own work," and for the
follow-ups that usually come after it.

---

## The short answer (60–90 seconds)

I published a study where I'd tested eleven predictors against about 600
observations, found nothing predictive, and written it up. Before I put it on
GitHub I committed the underlying data alongside the document — and then ran
the analysis fresh from the committed files to check the numbers matched what
I'd written.

Four of them didn't. Not conclusions, but figures: a median I'd quoted as
0.83% came out at 0.80%, a count I'd written as 43 of 231 turned out to be
121 of 615, a median drop of 2.79% was actually 2.66%. They were stale numbers
from earlier runs that never got updated when the pipeline changed.

None of them moved a single conclusion. But a reader who ran my own code would
have seen different numbers than my document claimed, and that's the kind of
mismatch that makes someone stop trusting everything else. So I reconciled all
four against what the code actually produced, and where a figure genuinely
differed between two collection methods, I reported both instead of picking one.

---

## The better story — the one where I was wrong twice

There's a second version that's more interesting, because I made the wrong call
first.

My write-up referenced a case where one observation had been measured over 116
periods and another over 2, as an illustration of why measurement window was
confounding my results. When I checked, neither committed data file could
produce a 116 — both capped at 32. So I concluded the figure was stale and
rewrote the sentence to use the 32-bar cap instead.

That was wrong. The 116 was real; it came from a third dataset I hadn't found
yet. When it turned up, it had exactly what I'd claimed — one leg at 103
periods, another at 116, no window cap at all, because that collection method
was different from the other two.

So I'd replaced a true statement with a weaker one, on reasoning that was sound
given what I could see. The lesson isn't "check harder" — I did check. It's
that a document and its data need to travel together. Every figure I couldn't
verify was one I'd eventually doubt for no reason, and the fix is committing
the data alongside the write-up from the start, not at publication.

---

## Follow-ups they're likely to ask

**"How did you catch it?"**
By making reproduction a requirement rather than an aspiration. I wrote a
notebook that runs the whole analysis from the committed files and prints every
headline figure. Then I compared its output against the document line by line.
The discrepancies surfaced immediately because I'd built something that would
surface them.

**"What if you'd published without checking?"**
Someone runs the notebook, sees 0.80% where the document says 0.83%, and now
has to decide whether that's a rounding difference or a sign the whole thing is
sloppy. They won't ask — they'll just close the tab. Small unexplained
mismatches cost more credibility than a stated limitation does.

**"Did it change your conclusions?"**
No, and that's worth saying plainly rather than overselling it. Four cosmetic
figures out of dozens. What it changed was whether the document could be
trusted by someone checking it. The cross-method comparison actually came out
*stronger* after reconciliation — I'd claimed two independent measurements both
gave 30%, and the real figures were 29.5% and 30.0%, with the median outcome
agreeing at 9.93% and 9.91%. I'd been underselling my own best evidence.

**"What would you do differently?"**
Commit the data on day one, not at publication. And record identifying
information on every hand-collected observation even when it seems irrelevant
to the hypothesis — I skipped timestamps on one dataset and it cost me the
ability to join it to the automated set for months, until the original files
turned up.

---

## The larger pattern worth connecting to

This is the same discipline as the main finding in the project. Five results
looked statistically significant and dissolved when I controlled for confounds —
one of them replicated at r = +0.70 and +0.66 across independent samples and was
*still* an artifact, because out-of-sample replication catches sampling noise
but not structural bias.

Checking my own document against my own data is the same instinct applied to the
write-up instead of the analysis. In both cases the thing that worked was direct
verification, not confidence that the process had been careful.

If they want a one-line version: I don't trust a number because I remember
producing it. I trust it because I can regenerate it.

---

## What not to say

- Don't apologise or frame it as a near-miss. It was routine verification that
  worked. Presenting it as a disaster averted invites the question of what else
  might be wrong.
- Don't oversell the stakes. Nobody was harmed by a 0.03% discrepancy in a
  personal research project. The transferable part is the habit, not the impact.
- Don't skip the part where I got the 116 wrong. That's the half that shows
  judgment under incomplete information, which is the actual skill being tested.
'''

pathlib.Path('NOTES-interview.md').write_text(content, encoding='utf-8')

gi = pathlib.Path('.gitignore')
lines = gi.read_text(encoding='utf-8').splitlines()
for name in ['NOTES-interview.md', 'repo_setup2.ipynb', 'reconcile_docs.ipynb',
             'fix_oi_section.ipynb', 'reframe_readme.ipynb']:
    if name not in [l.strip() for l in lines]:
        lines.append(name)
gi.write_text('\n'.join(lines) + '\n', encoding='utf-8')

print('wrote NOTES-interview.md')
print()
print(gi.read_text(encoding='utf-8'))
```
